## Data Cleaning
### Remove:
* Rows with missing values
* Numeric rows with iqr outliers
* Rows with invalid values based on columns
* Unique, constant, leakage and empty columns if any.

## Feature Engineering
Based on the available features for each dataset, create some new features from existing features like:
* Customer Age from Date of Birth
* Delivery delay from order and delivery date

```python
def remove_iqr_outliers(df, columns, multiplier=3.0):
    report = {}
    for col in columns:
        if col not in df.columns or not pd.api.types.is_numeric_dtype(df[col]):
            continue

        values = df[col].dropna()
        if values.empty:
            continue

        q1, q3 = values.quantile([0.25, 0.75])
        iqr = q3 - q1
        if not np.isfinite(iqr) or iqr == 0:
            continue

        lower = q1 - multiplier * iqr
        upper = q3 + multiplier * iqr
        before = len(df)
        df = df[df[col].isna() | df[col].between(lower, upper)].copy()

        report[col] = {
            "lower_bound": float(lower),
            "upper_bound": float(upper),
            "rows_removed": int(before - len(df)),
        }

    return df, report


def remove_empty_and_constant_columns(df, target):
    empty = [c for c in df.columns if c != target and df[c].isna().all()]
    df = df.drop(columns=empty)

    constant = [
        c for c in df.columns
        if c != target and df[c].nunique(dropna=False) <= 1
    ]
    df = df.drop(columns=constant)
    return df, empty, constant


def drop_all_missing_rows(df, report, stage):
    before = len(df)
    df = df.dropna(axis=0, how="any").copy()
    report[f"rows_removed_missing_{stage}"] = int(before - len(df))
    return df


def validate_binary_target(df, target, source_name):
    df[target] = pd.to_numeric(df[target], errors="coerce")
    before = len(df)
    df = df.dropna(subset=[target]).copy()
    removed = before - len(df)

    values = set(df[target].unique())
    if not values.issubset({0, 1}):
        raise ValueError(
            f"{source_name}: expected binary 0/1 target, found {sorted(values)}")
    df[target] = df[target].astype("int8")
    return df, int(removed)


def finish_dataset(df, report):
    df = drop_all_missing_rows(df, report, "final_features")

    df, empty, constant = remove_empty_and_constant_columns(df, "returned")
    report["empty_columns_removed_final"] = empty
    report["constant_columns_removed_final"] = constant

    if df.isna().any().any():
        raise RuntimeError(
            "Internal error: missing values remain after preprocessing.")
    if df.duplicated().any():
        df = df.drop_duplicates().copy()

    if df.empty:
        raise ValueError("No rows remain after preprocessing.")
    if df["returned"].nunique() != 2:
        raise ValueError(
            "Both return classes must remain after preprocessing.")

    report["rows_final"] = int(len(df))
    report["columns_final_including_target"] = int(len(df.columns))
    report["feature_count"] = int(len(df.columns) - 1)
    report["class_counts"] = {
        str(int(k)): int(v)
        for k, v in df["returned"].value_counts().sort_index().items()
    }
    report["return_rate"] = float(df["returned"].mean())
    report["missing_values_final"] = int(df.isna().sum().sum())
    report["duplicate_rows_final"] = int(df.duplicated().sum())
    return df, report


def preprocess_retail_return_risk(path):
    report = {"dataset": "retail_return_risk.csv", "target_source": "returned"}
    df = pd.read_csv(path)
    report["rows_original"] = int(len(df))
    report["duplicates_removed"] = int(df.duplicated().sum())
    df = df.drop_duplicates().replace([np.inf, -np.inf], np.nan).copy()

    nonnegative_cols = [
        "product_price", "discount_percent", "product_rating",
        "past_purchase_count", "past_return_rate",
        "session_length_minutes", "num_product_views",
    ]
    valid = pd.Series(True, index=df.index)
    for col in nonnegative_cols:
        if col in df:
            valid &= df[col].isna() | (df[col] >= 0)
    if "product_price" in df:
        valid &= df["product_price"].isna() | (df["product_price"] > 0)
    if "discount_percent" in df:
        valid &= df["discount_percent"].isna() | (
            df["discount_percent"] <= 100)
    if "past_return_rate" in df:
        valid &= df["past_return_rate"].isna() | (df["past_return_rate"] <= 1)
    if "product_rating" in df:
        valid &= df["product_rating"].isna() | (df["product_rating"] <= 5)
    if "customer_age" in df:
        valid &= df["customer_age"].isna(
        ) | df["customer_age"].between(13, 100)

    before = len(df)
    df = df.loc[valid].copy()
    report["invalid_domain_rows_removed"] = int(before - len(df))

    # Feature engineering.
    df["discount_amount"] = df["product_price"] * df["discount_percent"] / 100
    df["discounted_price"] = df["product_price"] - df["discount_amount"]

    dropped = [c for c in ["order_id"] if c in df.columns]
    df = df.drop(columns=dropped)
    report["unnecessary_or_leakage_columns_removed"] = dropped

    df, target_missing = validate_binary_target(
        df, "returned", "retail_return_risk.csv")
    report["invalid_target_rows_removed"] = target_missing

    df, outliers = remove_iqr_outliers(
        df,
        ["product_price", "past_purchase_count",
         "session_length_minutes", "num_product_views"],
        multiplier=3.0,
    )
    report["iqr_outliers"] = outliers

    df, empty, constant = remove_empty_and_constant_columns(df, "returned")
    report["empty_columns_removed"] = empty
    report["constant_columns_removed"] = constant

    df = drop_all_missing_rows(df, report, "final_features")
    return finish_dataset(df, report)


def preprocess_bads(path):
    report = {"dataset": "BADS_WS2021_known.csv", "target_source": "return"}
    df = pd.read_csv(path)
    report["rows_original"] = int(len(df))
    report["duplicates_removed"] = int(df.duplicated().sum())
    df = df.drop_duplicates().replace([np.inf, -np.inf], np.nan).copy()

    # Parse dates and create order-time features.
    for col in ["order_date", "user_dob", "user_reg_date", "delivery_date"]:
        df[col] = pd.to_datetime(df[col], errors="coerce")

    df["order_year"] = df["order_date"].dt.year
    df["order_month"] = df["order_date"].dt.month
    df["order_quarter"] = df["order_date"].dt.quarter
    df["order_day_of_week"] = df["order_date"].dt.dayofweek
    df["order_is_weekend"] = (
        df["order_date"].dt.dayofweek >= 5).astype("float")

    age = (df["order_date"] - df["user_dob"]).dt.days / 365.2425
    tenure = (df["order_date"] - df["user_reg_date"]).dt.days
    delivery_time = (df["delivery_date"] - df["order_date"]).dt.days

    df["customer_age_at_order"] = age.where(age.between(16, 100))
    df["account_tenure_days"] = tenure.where(tenure >= 0)
    df["delivery_time_days"] = delivery_time.where(delivery_time >= 0)

    # IDs are categorical identifiers, not continuous numeric quantities.
    for col in ["item_id", "brand_id"]:
        if col in df:
            df[col] = df[col].astype("string")

    # Basic domain validation.
    valid = pd.Series(True, index=df.index)
    if "item_price" in df:
        valid &= df["item_price"].isna() | (df["item_price"] > 0)
    before = len(df)
    df = df.loc[valid].copy()
    report["invalid_domain_rows_removed"] = int(before - len(df))

    df = df.rename(columns={"return": "returned"})
    df, target_missing = validate_binary_target(
        df, "returned", "BADS_WS2021_known.csv")
    report["invalid_target_rows_removed"] = target_missing

    dropped = [
        c for c in [
            "order_item_id", "user_id", "order_date", "delivery_date",
            "user_dob", "user_reg_date",
        ] if c in df
    ]
    df = df.drop(columns=dropped)
    report["unnecessary_or_leakage_columns_removed"] = dropped

    df, outliers = remove_iqr_outliers(
        df, ["item_price"], multiplier=3.0
    )
    report["iqr_outliers"] = outliers

    df, empty, constant = remove_empty_and_constant_columns(df, "returned")
    report["empty_columns_removed"] = empty
    report["constant_columns_removed"] = constant

    # This intentionally drops rows with missing DOB-derived age or any other
    # missing final feature, as requested (no imputation).
    df = drop_all_missing_rows(df, report, "final_features")
    return finish_dataset(df, report)


def preprocess_ecommerce(path):
    report = {"dataset": "Ecommerce_2000.csv", "target_source": "Returned"}
    df = pd.read_csv(path)
    report["rows_original"] = int(len(df))
    report["duplicates_removed"] = int(df.duplicated().sum())
    df = df.drop_duplicates().replace([np.inf, -np.inf], np.nan).copy()

    df["OrderDate"] = pd.to_datetime(df["OrderDate"], errors="coerce")
    dt = df["OrderDate"]
    df["OrderYear"] = dt.dt.year
    df["OrderMonth"] = dt.dt.month
    df["OrderQuarter"] = dt.dt.quarter
    df["OrderDayOfWeekNum"] = dt.dt.dayofweek
    df["OrderIsWeekend"] = (dt.dt.dayofweek >= 5).astype("float")

    # Validate source domains before engineering monetary features.
    valid = (
        (df["UnitPrice"] > 0)
        & (df["Quantity"] > 0)
        & df["DiscountPercent"].between(0, 100)
        & (df["ShippingDays"] >= 0)
    )
    before = len(df)
    df = df.loc[valid].copy()
    report["invalid_domain_rows_removed"] = int(before - len(df))

    target_map = {
        "yes": 1, "no": 0, "1": 1, "0": 0,
        "true": 1, "false": 0,
    }
    df["returned"] = (
        df["Returned"].astype(str).str.strip().str.lower().map(target_map)
    )
    before = len(df)
    df = df.dropna(subset=["returned"]).copy()
    report["invalid_target_rows_removed"] = int(before - len(df))
    df["returned"] = df["returned"].astype("int8")

    dropped = [
        c for c in [
            "OrderID", "CustomerID", "FirstName", "LastName",
            "OrderDate", "OrderDayOfWeek", "Returned",
        ] if c in df
    ]
    df = df.drop(columns=dropped)
    report["unnecessary_or_leakage_columns_removed"] = dropped

    df, outliers = remove_iqr_outliers(
        df, ["UnitPrice", "Quantity"], multiplier=4.0
    )
    report["iqr_outliers"] = outliers

    df, empty, constant = remove_empty_and_constant_columns(df, "returned")
    report["empty_columns_removed"] = empty
    report["constant_columns_removed"] = constant

    # Drop all rows with missing values in the final feature set.
    df = drop_all_missing_rows(df, report, "final_features")
    return finish_dataset(df, report)
```

```python
input_dir = "datasets"
output_dir = "processed_datasets"
os.makedirs(output_dir, exist_ok=True)

jobs = [
    ("Ecommerce_2000.csv", preprocess_ecommerce),
    ("retail_return_risk.csv", preprocess_retail_return_risk),
    ("BADS_WS2021_known.csv", preprocess_bads),
]

reports = []
for filename, processor in jobs:
    source = os.path.join(input_dir, filename)
    if not os.path.exists(source):
        raise FileNotFoundError(f"Missing input file: {source}")

    processed, report = processor(source)
    destination = os.path.join(output_dir, filename)
    processed.to_csv(destination, index=False)
    report["output_file"] = destination
    reports.append(report)

    print(
        f"{filename}: {report['rows_original']:,} -> "
        f"{report['rows_final']:,} rows; "
        f"{report['feature_count']} features; "
        f"return rate={report['return_rate']:.4f}"
    )

report_df = pd.DataFrame(reports)
report_df.to_csv(os.path.join(
    output_dir, "preprocessing_report.csv"), index=False)
with open(os.path.join(output_dir, "preprocessing_report.json"), "w") as f:
    json.dump(reports, f, indent=2, default=str)
```
